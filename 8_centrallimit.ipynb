# *Central Limit Theorem (CLT)*

The *Central Limit Theorem* states that when you repeatedly take random samples from any population (with finite mean and variance), the *distribution of the sample means* becomes approximately *normal*, even if the population itself is not normal.

---

## *Formal Statement*

If $X_1, X_2, \dots, X_n$ are independent and identically distributed (i.i.d.) random variables with:

- Mean: $E[X] = \mu$
- Variance: $\text{Var}(X) = \sigma^2$

Then the sampling distribution of the sample mean:

$ \bar{X} = \frac{1}{n} \sum_{i=1}^{n} X_i $

approaches a *normal distribution* as the sample size $n$ becomes large.

---

## *CLT Approximation*

For large $n$:

$ \bar{X} \sim N\left( \mu,\ \frac{\sigma^2}{n} \right) $

This can be standardized as:

$ Z = \frac{\bar{X} - \mu}{\sigma / \sqrt{n}} \approx N(0,1) $

---

## *Key Points*

- Works for *any population distribution* (skewed, uniform, unknown), as long as mean and variance exist.
- Sample size $n \ge 30$ is commonly considered “large enough”.
- Variability decreases as sample size increases:  
  $ \text{SD of } \bar{X} = \frac{\sigma}{\sqrt{n}} $

---

## *Interpretation*

The CLT explains why:

- Averages tend to be normally distributed  
- Many statistical procedures assume normality  
- Confidence intervals and hypothesis tests work even for non-normal data

---

## *Example*

Suppose you sample $n = 50$ people’s heights from a non-normal population with:

- Mean: $\mu = 165$
- Standard deviation: $\sigma = 10$

Then the sampling distribution of the mean is approximately:

$ \bar{X} \sim N\left(165,\ \frac{10^2}{50}\right) $

$ \bar{X} \sim N(165,\ 2) $

So most sample means will lie close to 165 with much smaller spread.

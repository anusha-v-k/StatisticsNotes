## Basics of Surveying in Data Analytics

Surveys are one of the most common methods to collect **primary data** in data analytics. They help gather opinions, behaviors, or factual information directly from people. Understanding survey concepts ensures that the data collected is reliable and useful for analysis.

---

### **1. What is a Survey?**
A **survey** is a structured method of collecting information from a targeted group (sample) of people using questions or prompts. The goal is to understand trends, opinions, or characteristics of a larger population.

---

## Key Terms and Definitions

### **2. Population**
The **entire group** you want to study.
- Example: All customers of a retail company.

### **3. Sample**
A **subset** taken from the population.
- Example: 500 customers selected from all customers.

---

### **4. Sampling Frame**
A **list of all members** in the population from which the sample is drawn.
- Example: Company’s customer database.

---

### **5. Sampling Methods**
How you choose the sample.

**a. Probability Sampling**  
Every member has a known chance of being selected.  
Examples:  
- Simple Random Sampling  
- Stratified Sampling  
- Systematic Sampling  
- Cluster Sampling

**b. Non-Probability Sampling**  
Not every member has a known chance.  
Examples:  
- Convenience Sampling  
- Judgmental Sampling  
- Snowball Sampling  
- Quota Sampling  

---

### **6. Survey Instruments**
Tools used to collect data.
- Questionnaires
- Online forms (Google Forms, SurveyMonkey)
- Interviews (phone, in-person)
- Observation sheets

---

### **7. Types of Survey Questions**
- **Open-Ended:** Free-text responses  
  *Example:* “What do you like about our service?”
- **Closed-Ended:** Fixed options  
  *Example:* MCQs, Yes/No, Rating scales
- **Likert Scale:** Measures attitude on a scale  
  (e.g., Strongly Agree → Strongly Disagree)
- **Semantic Differential Scale:** Opposite adjectives  
  (e.g., “Fast — Slow”)

---

### **8. Response Rate**
Percentage of people who responded out of those who were invited.  
- Formula:  
  **Response Rate = (Number of Responses ÷ Total Invitations) × 100**

---

### **9. Bias**
Errors that cause results to be inaccurate.
- **Sampling Bias:** Sample does not represent population  
- **Response Bias:** People answer dishonestly  
- **Non-Response Bias:** Certain groups don’t respond  

---

### **10. Reliability and Validity**
- **Reliability:** Consistency of results if repeated  
- **Validity:** Whether the survey measures what it intends to measure  

---

### **11. Pilot Survey**
A **small trial run** of the survey to identify issues before full deployment.

---

### **12. Survey Modes**
- Online surveys  
- Telephone surveys  
- Face-to-face interviews  
- Mail or paper surveys  

---

### **13. Data Cleaning in Surveys**
Once data is collected, you may need to:
- Remove incomplete responses  
- Standardize categories  
- Handle missing values  
- Decode text responses  

---

Surveys form the backbone of many analytics projects, and understanding these fundamentals helps in designing better questionnaires and interpreting data accurately.


## Parameter vs Statistic

### **Population**
A *population* is the entire group of individuals, items, or observations that we want to study.  
Examples: all students in a university, all cars produced in a factory, all customers of a store.

### **Sample**
A *sample* is a subset of the population selected for analysis.  
Examples: 200 students selected from the university, 50 cars from the factory, 100 customers surveyed.

---

## **Parameter**
A **parameter** is a numerical value that describes a characteristic of a *population*.  
It is:
- Fixed (but usually unknown)
- Represents the true value for the whole population

**Examples:**
- Population mean height (μ)
- Population standard deviation (σ)
- Population proportion of defected items (P)

Because measuring the entire population is often impractical, parameters are rarely known.

---

## **Statistic**
A **statistic** is a numerical value that describes a characteristic of a *sample*.  
It is:
- Calculated from sample data
- Used to estimate the population parameter
- Can change from sample to sample

**Examples:**
- Sample mean (𝑥̄)
- Sample standard deviation (s)
- Sample proportion (p̂)

---

## **Relationship Between Parameter and Statistic**
- A **statistic** is used as an **estimate** for a **parameter**.
- Example:  
  If you cannot measure the mean height of all people in a city (population),  
  you measure a sample and use its mean (statistic) to estimate μ (parameter).

---

### **Summary Table**

| Concept      | Based On     | Symbol Example | Nature | Example |
|--------------|--------------|----------------|--------|---------|
| **Parameter** | Population    | μ, σ, P        | Fixed, usually unknown | True mean income of all citizens |
| **Statistic** | Sample        | 𝑥̄, s, p̂       | Varies sample to sample | Mean income of surveyed 200 people |



### Vague Wording in Surveys

Vague wording refers to survey questions that are unclear, ambiguous, or open to multiple interpretations. When respondents do not fully understand what is being asked, their answers may become inaccurate or inconsistent.

#### Why Vague Wording Is a Problem?
- It reduces the reliability of responses  
- It introduces measurement error  
- Respondents may guess the meaning, leading to biased results  
- Different respondents may interpret the same question differently  

#### Examples of Vague Survey Questions
- **"Do you exercise regularly?"**  
  - *"Regularly"* is unclear—does it mean daily, weekly, or monthly?

- **"Do you earn a high income?"**  
  - *"High"* is subjective and varies by person.

- **"Do you often eat outside?"**  
  - *"Often"* can mean different things to different people.

#### Improved (Clear) Versions
- "How many days per week do you exercise?"  
- "What is your monthly income (in ₹)?"  
- "How many times per month do you eat outside?"

#### Best Practices to Avoid Vague Wording
- Use specific time frames  
- Define key terms clearly  
- Use quantifiable options  
- Avoid subjective words like *often, regularly, many, high, frequently*  



## Recall Bias

- Recall bias happens when respondents do not remember past events accurately while answering a survey, especially when the recall period is long or the event was minor.

- Example

  - Asking: “How many times did you eat fast food in the last 6 months?”
   → People may underestimate or overestimate because they cannot remember precisely.

## **Blinding and Double-Blinding in Experiments**

### **Blinding**
Blinding is a technique used in experiments to reduce **bias** by ensuring that participants **do not know** which treatment group they belong to.  
The goal is to prevent their expectations, beliefs, or behaviors from influencing the results.

- **Example:**  
  In a clinical trial, participants may receive either a real drug or a placebo.  
  If participants do not know which one they received, the study is **single-blind**.

### **Double-Blinding**
Double-blinding goes a step further.  
In a double-blind experiment, **both**:
- the participants, and  
- the researchers who interact with them or measure outcomes  

**do not know** who received the treatment and who received the placebo.

This prevents:
- participants' expectations from influencing outcomes  
- researchers’ expectations from affecting observations, measurement, or behavior

- **Example:**  
  In a drug study, neither the patients nor the doctors administering the treatment know who is receiving the real drug and who is receiving the placebo.  
  This reduces **observer bias** and **participant bias**.

### **Why Blinding Matters**
- Prevents intentional or unintentional influence  
- Ensures objective measurement  
- Improves the credibility and validity of results  

### **Summary Table**

| Type of Blinding | Participants Know? | Researchers Know? | Purpose |
|------------------|--------------------|--------------------|---------|
| **Single-Blind** | ❌ No | ✔️ Yes | Prevent participant bias |
| **Double-Blind** | ❌ No | ❌ No | Prevent both participant and researcher bias |

## What Is Hypothesis Testing?

- Hypothesis testing is a structured procedure in statistics used to decide whether there is enough evidence in a sample of data to believe something is true for the entire population.

#### Null Hypothesis (H₀)
- A statement of no effect, no difference, or status quo.
  - Example:

    - “The new ad campaign does NOT increase sales.”

    - “The mean weight of packets is still 500g.”

#### Alternative Hypothesis (H₁ or Ha)
- What you want to test or prove.
  - Example:

    - “The new ad campaign increases sales.”

    - “The mean weight is not 500g anymore.”

## ⭐ Significance Level (α)

The **significance level**, denoted by **α**, is a threshold used in hypothesis testing to decide whether the evidence from the sample is strong enough to reject the **null hypothesis (H₀)**.

### 🔍 What α Represents
- **α is the probability of making a Type I error**  
  (rejecting H₀ when it is actually true — a “false positive”).

### 🎯 Common Values of α
- **0.05** → Most commonly used  
- **0.01** → Very strict (strong evidence required)  
- **0.10** → More lenient

### 📌 How α Works in Decision Making
- If **p-value < α** → Reject H₀ (evidence is statistically significant)  
- If **p-value ≥ α** → Fail to reject H₀ (evidence is not strong enough)

### 🧠 Interpretation
If **α = 0.05**, it means:
> You accept a **5% chance** of wrongly rejecting a true null hypothesis.

### 📊 In a Two-Tailed Test
The α value is split across both tails:
- Left tail: **α/2**
- Right tail: **α/2**

### 📘 Example
If α = 0.05 and your test gives **p = 0.03**:
- Since **0.03 < 0.05**, you **reject H₀**.

The significance level sets the standard for how much risk of error you are willing to accept before concluding the effect is real.


## 🎯 Type I and Type II Errors in Hypothesis Testing

When we make decisions using hypothesis tests, two kinds of mistakes are possible:

---

## ❌ Type I Error (False Positive)

- Occurs when you **reject the null hypothesis (H₀)** even though it is actually **true**.
- You are detecting an effect **that doesn’t exist**.

### 🔥 Probability of Type I Error = **α (significance level)**

Example:  
You conclude that a new drug works (reject H₀), but in reality, it does **not** work.

---

## ❌ Type II Error (False Negative)

- Occurs when you **fail to reject the null hypothesis (H₀)** even though the **alternative hypothesis (H₁)** is true.
- You **miss an effect that actually exists**.

### 🔥 Probability of Type II Error = **β**

Example:  
You conclude that a machine produces packets with mean weight = 500g (fail to reject H₀), but in reality, the true mean has changed.

---

## ⚡ Power of a Test

The **power** of a test = **1 − β**  
It tells you the probability of **correctly rejecting H₀** when H₁ is true.

High power = low chance of Type II error.

---

## 📘 Summary Table

| Decision / Truth        | H₀ True            | H₀ False                 |
|-------------------------|--------------------|---------------------------|
| **Reject H₀**           | Type I Error (α)   | Correct Decision          |
| **Fail to Reject H₀**   | Correct Decision   | Type II Error (β)         |

---

## 🧠 Easy Way to Remember
- **Type I = False Alarm** (thinking an effect exists when it doesn’t)  
- **Type II = Missed Detection** (failing to notice an effect that exists)

- The p-value depends on the test statistic and the distribution used in the hypothesis test.

### 🔹 Z-Test: Explanation, When to Use, and Formula

A **Z-test** is a statistical hypothesis test used to determine whether there is a significant difference between a sample statistic and a population parameter, assuming the sampling distribution is **approximately normal**.

---

### ✅ **When to Use a Z-Test**

Use a Z-test when:

1. **Population standard deviation ($\sigma$) is known**  
2. **Sample size is large** ($n \ge 30$), OR the population is normally distributed  
3. You are testing for:
   - Population mean  
   - Difference between two population means  
   - Population proportion  

---

### 📌 **Z-Test for a Single Mean**

You test:

- **Null hypothesis:**  
  $$H_0: \mu = \mu_0$$

- **Alternative hypothesis:**  
  $$H_1: \mu \ne \mu_0 \quad \text{(two-tailed)}$$  
  or  
  $$H_1: \mu > \mu_0 \quad \text{(right-tailed)}$$  
  $$H_1: \mu < \mu_0 \quad \text{(left-tailed)}$$

---

### 🔢 **Z-Statistic Formula**

If the population standard deviation is known:

$$
Z = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}}
$$

Where:  
- $\bar{X}$ = sample mean  
- $\mu_0$ = hypothesized population mean  
- $\sigma$ = population standard deviation  
- $n$ = sample size  

---

### 📝 **Decision Rule**

- Compute the **Z value**
- Compare with critical Z from standard normal table  
  (e.g., ±1.96 for 5% significance, two-tailed)
- Or compute p-value:
  - If **p ≤ \alpha**, reject $H_0$  
  - If **p > \alpha**, fail to reject $H_0$

---

### ✔️ Summary

A **Z-test** is best used when the sample is large or $\sigma$ is known, making the sampling distribution of the mean normal. It is widely used in quality control, A/B testing, and comparing sample values to a population benchmark.

### Confidence Level and Confidence Interval

---

### Confidence Level

The **confidence level** is the proportion of times the confidence interval would contain the true population parameter if we repeated the sampling process many times.

Common confidence levels are:

- 90%
- 95%
- 99%

A 95% confidence level means that 95% of the confidence intervals created from repeated random samples will contain the true population mean.

---

### Confidence Interval (CI)

A **confidence interval** is a range of values estimated from a sample that is likely to contain the true population parameter.

For a population mean with known standard deviation, the confidence interval is:

$$
\bar{X} \pm Z_{\alpha/2} \left( \frac{\sigma}{\sqrt{n}} \right)
$$

Where:

- $\bar{X}$ = sample mean  
- $\sigma$ = population standard deviation  
- $n$ = sample size  
- $Z_{\alpha/2}$ = critical Z-value for the confidence level  
  - 90% → 1.645  
  - 95% → 1.96  
  - 99% → 2.576

---

### Example

Suppose  
$\bar{X} = 50$,  
$\sigma = 10$,  
$n = 64$,  
confidence level = 95%.

Step 1:  
$$
Z_{0.025} = 1.96
$$

Step 2:  
$$
\text{Margin of Error} = 1.96 \cdot \frac{10}{\sqrt{64}}
$$

Step 3:  
$$
\text{CI} = 50 \pm 2.45 = (47.55,\ 52.45)
$$

Interpretation:  
We are 95% confident that the true population mean lies between **47.55** and **52.45**.

## What is Entropy? (Simple Explanation)

- Entropy is a measure of uncertainty, randomness, or impurity in a system.

## 🔢 Entropy Formula (Information Theory)

For a dataset with class probabilities $p_1, p_2, \ldots, p_k$, the entropy is defined as:

$$
\text{Entropy} = -\sum_{i=1}^{k} p_i \log_2(p_i)
$$

### 📌 Special Case: Binary Classification
If the two class probabilities are $p$ and $1 - p$:

$$
\text{Entropy} = - \left[ p \log_2(p) + (1 - p)\log_2(1 - p) \right]
$$

### ✔ Notes
- Entropy = 0 when the dataset is perfectly pure.  
- Entropy is maximum when both classes are equally likely.  
- Log base 2 is used to measure entropy in **bits**.


## 🎯 Entropy Calculation (6 Black Balls, 4 White Balls)

Total balls:  
- Black = 6  
- White = 4  
- Total = 10  

### Step 1: Compute class probabilities
- Probability of Black:  
  $p_{\text{black}} = \frac{6}{10} = 0.6$

- Probability of White:  
  $p_{\text{white}} = \frac{4}{10} = 0.4$

### Step 2: Apply entropy formula

$$
\text{Entropy} = - \left[ 
p_{\text{black}} \log_2(p_{\text{black}}) \;+\; 
p_{\text{white}} \log_2(p_{\text{white}})
\right]
$$

Substitute values:

$$
\text{Entropy} = - \left[
0.6 \log_2(0.6) + 0.4 \log_2(0.4)
\right]
$$

### Step 3: Compute logarithms
- $\log_2(0.6) \approx -0.737$
- $\log_2(0.4) \approx -1.322$

### Step 4: Final entropy value

$$
\text{Entropy} 
= - \left[
0.6(-0.737) + 0.4(-1.322)
\right]
$$

$$
= - \left[
-0.4422 - 0.5288
\right]
$$

$$
= 0.971
$$

### ✅ Final Answer
**Entropy = 0.971 bits**


## 🎯 Entropy Calculation (5 Black Balls, 5 White Balls)

Total balls:  
- Black = 5  
- White = 5  
- Total = 10  

### Step 1: Compute class probabilities
- Probability of Black:  
  $p_{\text{black}} = \frac{5}{10} = 0.5$

- Probability of White:  
  $p_{\text{white}} = \frac{5}{10} = 0.5$

### Step 2: Apply entropy formula

$$
\text{Entropy} = - \left[ 
p_{\text{black}} \log_2(p_{\text{black}}) \;+\; 
p_{\text{white}} \log_2(p_{\text{white}})
\right]
$$

Substitute values:

$$
\text{Entropy} = - \left[
0.5 \log_2(0.5) + 0.5 \log_2(0.5)
\right]
$$

### Step 3: Compute logarithms
- $\log_2(0.5) = -1$

### Step 4: Final entropy value

$$
\text{Entropy} = - \left[ 0.5(-1) + 0.5(-1) \right] 
= -(-1) 
= 1
$$

### ✅ Final Answer
**Entropy = 1 bit** (maximum uncertainty for binary classes)


## 🌳 Decision Tree (ID3) Example — Tennis Dataset

### Sample Dataset

| Day | Outlook  | Temperature | Humidity | Wind   | PlayTennis |
|-----|----------|-------------|---------|--------|------------|
| 1   | Sunny    | Hot         | High    | Weak   | No         |
| 2   | Sunny    | Hot         | High    | Strong | No         |
| 3   | Overcast | Hot         | High    | Weak   | Yes        |
| 4   | Rain     | Mild        | High    | Weak   | Yes        |
| 5   | Rain     | Cool        | Normal  | Weak   | Yes        |
| 6   | Rain     | Cool        | Normal  | Strong | No         |
| 7   | Overcast | Cool        | Normal  | Strong | Yes        |
| 8   | Sunny    | Mild        | High    | Weak   | No         |
| 9   | Sunny    | Cool        | Normal  | Weak   | Yes        |
| 10  | Rain     | Mild        | Normal  | Weak   | Yes        |
| 11  | Sunny    | Mild        | Normal  | Strong | Yes        |
| 12  | Overcast | Mild        | High    | Strong | Yes        |
| 13  | Overcast | Hot         | Normal  | Weak   | Yes        |
| 14  | Rain     | Mild        | High    | Strong | No         |

---

### Step 1: Entropy of Entire Dataset

- Yes = 9, No = 5  
- Total = 14  

$$
p_\text{Yes} = 9/14 = 0.643,\quad p_\text{No} = 5/14 = 0.357
$$

$$
\text{Entropy}(S) = -0.643\log_2 0.643 - 0.357\log_2 0.357
$$

$$
\text{Entropy}(S) = 0.942 \text{ bits}
$$

---

### Step 2: Information Gain for “Outlook”

- Outlook values: Sunny (5), Overcast (4), Rain (5)

**Entropy for each value**:

- Sunny: 2 Yes, 3 No →  
$$
\text{Entropy(Sunny)} = 0.971
$$
- Overcast: 4 Yes, 0 No →  
$$
\text{Entropy(Overcast)} = 0
$$
- Rain: 3 Yes, 2 No →  
$$
\text{Entropy(Rain)} = 0.971
$$

**Weighted Entropy for Outlook**:

$$
\text{Entropy}_{\text{Outlook}} = \frac{5}{14}\cdot0.971 + \frac{4}{14}\cdot0 + \frac{5}{14}\cdot0.971 = 0.694
$$

**Information Gain**:

$$
\text{Gain(Outlook)} = \text{Entropy}(S) - \text{Entropy}_{\text{Outlook}} = 0.942 - 0.694 = 0.248
$$

---

### Step 3: Information Gain for Other Attributes

| Attribute    | Gain    |
|-------------|---------|
| Temperature | 0.031   |
| Humidity    | 0.153   |
| Wind        | 0.050   |

✅ **Root Node = Outlook** (highest gain)

---

### Step 4: Split by Outlook

- **Outlook = Overcast** → Yes (pure)  
- **Outlook = Sunny** → further split using attributes (Humidity, Wind, Temperature)  
- **Outlook = Rain** → further split using attributes

**Repeat steps recursively** until all nodes are **pure or no attributes remain**.

---

### ✅ Summary

1. Compute **Entropy(S)**  
2. Compute **Information Gain** for all attributes  
3. Choose **attribute with highest Gain** as root/split  
4. Split dataset and repeat recursively  
5. Stop when nodes are **pure** or **no attributes left**

This is the **ID3 Decision Tree algorithm** in action.

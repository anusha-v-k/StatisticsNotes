# Conditional Probability

Conditional probability tells us the probability of an event happening **given** that another event has already occurred.

---

## **Definition**

If:

- $A$ = event of interest  
- $B$ = event known to have occurred  

Then the conditional probability of $A$ given $B$ is:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

(valid only when $P(B) > 0$)

---

## **Example 1: Cards**

A deck has 52 cards.

Let  
- $A$ = drawing a King  
- $B$ = drawing a Heart  

There is only 1 King of Hearts and 13 Hearts total.

$$
P(A \mid B) = \frac{1}{13}
$$

---

## **Example 2: Students**

A class has 40 students:

|                | Male | Female | Total |
|----------------|------|--------|-------|
| Passed         | 12   | 16     | 28    |
| Failed         |  6   |  6     | 12    |
| **Total**      | 18   | 22     | 40    |

Find $P(\text{Male} \mid \text{Passed})$.

$$
P(\text{Male} \mid \text{Passed}) = \frac{12}{28} = \frac{3}{7}
$$

---

## **Intuition**

Conditional probability **restricts the sample space** to the condition that is already known to be true.

---

## **Bayes’ Theorem (related)**

$$
P(A \mid B) = \frac{P(B \mid A)\, P(A)}{P(B)}
$$

---

## **Practice Questions**

1. A box has 5 red and 7 blue balls. One ball is drawn and is blue.  
   What is the probability the next ball is red (without replacement)?

2. From 100 patients:  
   - 60 have fever  
   - 30 have cough  
   - 25 have both  

   Find $P(\text{fever} \mid \text{cough})$.


# Bayes' Theorem

Bayes’ Theorem is used to **reverse conditional probabilities**.  
It allows us to find $P(A \mid B)$ when we know $P(B \mid A)$.

---

## **Formula**

If $A$ and $B$ are two events with $P(B) > 0$, then:

$$
P(A \mid B) = \frac{P(B \mid A)\, P(A)}{P(B)}
$$

---

## **Expanded Form (when $B$ can occur in multiple ways)**

If events $A_1, A_2, \dots, A_n$ form a partition of the sample space, then:

$$
P(A_i \mid B)
= \frac{P(B \mid A_i)\, P(A_i)}
{\sum_{j=1}^{n} P(B \mid A_j)\, P(A_j)}
$$

---

## **Intuition**

Bayes’ theorem updates the probability of an event **after new evidence arrives**.

Think of it as:

$$
\text{Posterior} = \frac{\text{Likelihood} \times \text{Prior}}{\text{Total Evidence}}
$$

---

## **Example**

Suppose:

- $1\%$ of people have a certain disease  
- A test is $99\%$ accurate when a person has the disease  
- The test gives a false positive $5\%$ of the time

Find the probability that a person actually has the disease **given the test is positive**.

Let:  
- $A$ = person has disease  
- $B$ = test is positive  

Given:  
- $P(A)=0.01$  
- $P(B \mid A)=0.99$  
- $P(B \mid A^c)=0.05$

Total probability of a positive test:

$$
P(B) = P(B \mid A)P(A) + P(B \mid A^c)P(A^c)
$$

$$
P(B) = (0.99)(0.01) + (0.05)(0.99)
$$

Then Bayes’ theorem:

$$
P(A \mid B) = \frac{P(B \mid A)P(A)}{P(B)}
$$

---

## **Interpretation**

Even after a positive test, the chance of having the disease is **much lower than expected**, because the disease is very rare.

---

## **Practice Question**

A factory has 3 machines:

- $M_1$ produces $40\%$ of items, defect rate $3\%$  
- $M_2$ produces $35\%$ of items, defect rate $5\%$  
- $M_3$ produces $25\%$ of items, defect rate $2\%$

An item is found defective.  
Find $P(M_2 \mid \text{defective})$.



# Bayes' Theorem: Explanation of Each Term

Bayes' theorem can be written as:

$$
\text{Posterior} = \frac{\text{Likelihood} \times \text{Prior}}{\text{Total\ Evidence}}
$$

---

## **1. Prior**

- The **prior probability** represents your belief about an event *before* seeing the evidence.
- It is written as:

  $$
  \text{Prior} = P(A)
  $$

- Example: If $1\%$ of people have a disease, then:

  $$
  P(\text{Disease}) = 0.01
  $$

---

## **2. Likelihood**

- The **likelihood** is the probability of the evidence assuming the event is true.
- It is written as:

  $$
  \text{Likelihood} = P(B \mid A)
  $$

- Example: If a test detects the disease correctly $99\%$ of the time:

  $$
  P(\text{Positive} \mid \text{Disease}) = 0.99
  $$

---

## **3. Total Evidence (Normalization Term)**

- The **total evidence** is the probability of observing the evidence from *all possible causes*.
- Written as:

  $$
  \text{Total\ Evidence} = P(B)
  $$

- Computed using the law of total probability:

  $$
  P(B) = P(B \mid A)P(A) + P(B \mid A^c)P(A^c)
  $$

- Example:

  $$
  P(B) = 0.99(0.01) + 0.05(0.99)
  $$

---

## **4. Posterior**

- The **posterior probability** is the updated probability of the event *after* seeing the evidence.
- It is written as:

  $$
  \text{Posterior} = P(A \mid B)
  $$

This is the final answer you compute after adjusting your belief using the evidence.

---

## **Putting It All Together**

Using Bayes’ theorem:

$$
P(A \mid B) = \frac{P(B \mid A)P(A)}{P(B)}
$$

---

## **Quick Numerical Example**

Given:

- $P(\text{Disease}) = 0.01$
- $P(\text{Positive} \mid \text{Disease}) = 0.99$
- $P(\text{Positive} \mid \text{No Disease}) = 0.05$

Compute total evidence:

$$
P(B) = 0.99(0.01) + 0.05(0.99) = 0.0594
$$

Posterior:

$$
P(\text{Disease} \mid \text{Positive})
= \frac{0.0099}{0.0594}
= 0.1667
$$

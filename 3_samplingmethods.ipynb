## Sampling Methods with Examples

Sampling is the process of selecting a smaller group (sample) from a larger group (population) to conduct analysis. Choosing the right sampling method ensures accurate and unbiased results.

---

## **1. Probability Sampling**
Every member of the population has a known and non-zero chance of being selected.

### **a. Simple Random Sampling**
Each individual has an equal chance of being selected.
- **Example:** You have 1,000 customers. You randomly pick 100 using a random number generator.

### **b. Systematic Sampling**
Select every *k-th* member in a list.
- **Example:** From a list of 5,000 employees, you choose every 50th employee.

### **c. Stratified Sampling**
Population is divided into groups (strata), then samples are taken from each group proportionally.
- **Example:** A university divides students by year (1st, 2nd, 3rd, 4th year) and samples 50 students from each group.

### **d. Cluster Sampling**
Population is divided into clusters, a few clusters are selected randomly, and **all individuals** in selected clusters are surveyed.
- **Example:** A city has 20 schools. You randomly pick 3 schools and survey every student in those schools.

---

## **2. Non-Probability Sampling**
Not every member has a known chance of selection. Easier, but may introduce bias.

### **a. Convenience Sampling**
Sample is chosen based on ease of access.
- **Example:** A survey team stands in a mall and asks the first 100 people they meet.

### **b. Judgmental (Purposive) Sampling**
Researcher selects people based on their judgment or expertise.
- **Example:** A manager selects 10 “most experienced” employees for a skill assessment.

### **c. Snowball Sampling**
Existing participants recruit future participants—useful for hard-to-reach groups.
- **Example:** Studying people with a rare medical condition. One participant leads the researcher to others.

### **d. Quota Sampling**
Population is divided into groups, and a fixed number (quota) is taken from each group, but selection within groups is non-random.
- **Example:** You need 50 men and 50 women for a survey, and you pick them based on who is available.

Sampling method selection depends on time, cost, population structure, and level of accuracy needed.


## Cons of Stratified Sampling

While stratified sampling improves accuracy by ensuring representation from each subgroup (strata), it also comes with several limitations:

### **1. Requires Detailed Population Information**
To create strata, you need prior knowledge about the population (e.g., age groups, income levels).  
If this information is missing or inaccurate, the method may fail.

### **2. Time-Consuming and Complex**
Identifying appropriate strata and dividing the population correctly takes more effort compared to simple random sampling.

### **3. Not Suitable for All Populations**
If the population cannot be clearly divided into distinct, non-overlapping strata, stratified sampling becomes impractical.

### **4. Risk of Misclassification**
Members may be placed into wrong strata due to errors, biases, or unclear criteria—leading to inaccurate results.

### **5. Increased Cost**
More steps (data collection, grouping, proportional allocation) can mean higher administrative and operational costs.

### **6. Inefficiency if Strata Are Not Homogeneous**
Stratified sampling works best when individuals within each stratum are similar.  
If strata are poorly defined or very diverse internally, it reduces the method’s benefits.

---

Despite these cons, stratified sampling is widely used when accuracy and subgroup representation are important.


## Disproportionate Stratified Sampling

In **disproportionate stratified sampling**, the sample size taken from each stratum **is not proportional** to the size of each group in the population.  
This method is used when:
- Certain groups are more important for the study  
- Some strata are small and need to be oversampled for better analysis  
- You want equal representation across groups  

---

## Example: Male (10) and Female (20)

**Population**
- Male = 10  
- Female = 20  
- Total = 30  

Suppose we want a total sample of **10 people**, but we choose **equal allocation** instead of proportional.

### **Proportional Allocation (for comparison)**
- Male proportion = 10/30 → expected = 3 or 4  
- Female proportion = 20/30 → expected = 6 or 7  

---

## **Disproportionate Allocation Example**
Let's take **5 males** and **5 females**, even though their population sizes differ.

| Group   | Population Size | Proportion in Population | Sample Size (Disproportionate) |
|---------|------------------|---------------------------|--------------------------------|
| Male    | 10               | 33%                       | **5** |
| Female  | 20               | 67%                       | **5** |
| **Total** | **30**         | —                         | **10** |

Here, males are **oversampled** (they get 5 instead of ~3) and females are **undersampled** (they get 5 instead of ~7).

---

## When Disproportionate Sampling Is Used

1. **Small but important strata**  
   - A small subgroup might need extra representation to ensure reliable analysis.

2. **Ensuring equal representation**  
   - Sometimes each group gets the same sample size for fairness in comparison.

3. **Cost or time constraints**  
   - Easier to survey equal numbers from each group.

4. **Improving precision for specific strata**  
   - Oversampling a subgroup can help get more accurate insights for that group.

---

## Key Point
Because the sample distribution does NOT reflect the population distribution, **weights** may be needed during analysis to avoid bias.

---

Disproportionate stratification is powerful when used intentionally and interpreted correctly.


## Multistage Sampling

**Multistage sampling** is a complex form of cluster sampling where the researcher selects samples in **multiple stages**, gradually narrowing down the population.  
Instead of selecting final respondents directly, you move step-by-step through levels (stages) of groups.

It is useful when:
- The population is large and geographically spread out  
- A complete list of all individuals is not available  
- Time and cost need to be minimized  

---

## How Multistage Sampling Works

### **Stage 1:** Select large clusters  
### **Stage 2:** Select smaller sub-clusters within those clusters  
### **Stage 3:** Select individuals within sub-clusters  
(You can have 2, 3, 4… stages depending on study design)

---

## Example 1: Household Survey in a Country

**Goal:** Survey 200 households across the country.

### **Stage 1:** Select states  
Randomly pick 4 out of 28 states.

### **Stage 2:** Select districts  
From each selected state, randomly pick 2 districts.

### **Stage 3:** Select villages  
From each district, randomly pick 3 villages.

### **Stage 4:** Select households  
From each selected village, randomly select 10 households.

This stepwise approach makes data collection easier and cheaper.

---

## Example 2: Student Performance Study

**Goal:** Study 100 students in a large school system.

### **Stage 1:** Select schools  
Pick 5 random schools from the district.

### **Stage 2:** Select classes  
From each school, pick 2 classes at random.

### **Stage 3:** Select students  
From each class, select 10 students.

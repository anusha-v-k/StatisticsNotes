# *Empirical Rule (68–95–99.7 Rule)*

The *Empirical Rule* applies to *bell-shaped, normal distributions*.  
It states that most of the data falls within 1, 2, and 3 standard deviations of the mean.

---

## *Rule Summary*

- *68%* of the data lie within *$1\sigma$* of the mean  
  $ \mu - \sigma \le X \le \mu + \sigma $

- *95%* of the data lie within *$2\sigma$* of the mean  
  $ \mu - 2\sigma \le X \le \mu + 2\sigma $

- *99.7%* of the data lie within *$3\sigma$* of the mean  
  $ \mu - 3\sigma \le X \le \mu + 3\sigma $

---

## *Graph Meaning*

For a normal distribution:

- $68\%$ of values are close to the mean  
- $95\%$ of values are moderately far  
- $99.7\%$ of values are far but still within the typical spread

---

## *Example*

If a dataset is normally distributed with:

- Mean: $\mu = 100$  
- Standard deviation: $\sigma = 10$

Then:

- *68%* of data lie between  
  $100 - 10 = 90$ and $100 + 10 = 110$

- *95%* lie between  
  $100 - 20 = 80$ and $100 + 20 = 120$

- *99.7%* lie between  
  $100 - 30 = 70$ and $100 + 30 = 130$

# *Chebyshev's Theorem*

Chebyshev's Theorem (or Chebyshev's Inequality) states that for *any distribution*, regardless of shape, at least a certain proportion of the data will lie within $k$ standard deviations from the mean, where $k > 1$.

---

## *Formula*

For any dataset with mean $\mu$ and standard deviation $\sigma$:

$ P\left( |X - \mu| < k\sigma \right) \ge 1 - \frac{1}{k^2} $

This means that the proportion of data within $k$ standard deviations from the mean is *at least*:

$ 1 - \frac{1}{k^2} $

---

## *Interpretation*

- Works for *any* distribution: symmetrical, skewed, unknown, etc.  
- Guarantees a *minimum* proportion of data in the interval.  
- Larger $k$ → more data captured.

---

## *Common Values*

- For $k = 2$:  
  $ 1 - \frac{1}{2^2} = 1 - \frac{1}{4} = 0.75 $  
  *At least 75%* of observations lie within $2\sigma$ of the mean.

- For $k = 3$:  
  $ 1 - \frac{1}{3^2} = 1 - \frac{1}{9} = 0.888... $  
  *At least 88.9%* of observations lie within $3\sigma$ of the mean.

---

## *Interval Form*

Chebyshev’s Theorem guarantees:

$ \mu - k\sigma \le X \le \mu + k\sigma $

At least  
$ 1 - \frac{1}{k^2} $  
of the data falls inside this interval.

---

## *Example*

Suppose a dataset has:

- Mean: $\mu = 50$  
- Standard deviation: $\sigma = 8$  
- Choose $k = 2$

Interval:

$ 50 - 2(8) = 34 $  
$ 50 + 2(8) = 66 $

Chebyshev guarantees that *at least 75%* of the data lies between *34* and *66*.

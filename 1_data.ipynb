## Data Types in Statistics

Understanding data types is essential because they determine which statistical methods and visualizations are appropriate. Statistical data is broadly classified into **Quantitative** and **Qualitative** types.

---

### **1. Qualitative (Categorical) Data**
These represent categories or labels rather than numerical values.

#### **a. Nominal Data**
- Categories with **no natural order**
- Used for labeling or naming
- Examples: Gender (Male/Female), Colors (Red/Blue/Green), Country

#### **b. Ordinal Data**
- Categories with a **meaningful order**, but differences between categories are not measurable
- Examples: Customer satisfaction levels (Low/Medium/High), Education levels, Rankings

---

### **2. Quantitative (Numerical) Data**
These represent measurable quantities and support arithmetic operations.

#### **a. Discrete Data**
- Consists of **countable** values
- No decimals/fractions in most scenarios
- Examples: Number of students, Number of products sold

#### **b. Continuous Data**
- Can take **any value within a range**
- Decimals/fractions possible
- Examples: Height, Weight, Temperature, Time

---

### **Summary Table**

| Type | Subtype | Description | Examples |
|------|---------|-------------|----------|
| **Qualitative** | Nominal | Categories without order | Eye color, Gender |
| | Ordinal | Categories with order | Ratings, Education level |
| **Quantitative** | Discrete | Countable values | Number of cars, Calls per day |
| | Continuous | Measurable values in a range | Height, Temperature |

---

Understanding these data types helps you choose the right statistical tools, charts, and models for analysis.
